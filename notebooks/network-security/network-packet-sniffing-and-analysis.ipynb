# network packet sniffing and analysis

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
!pip install scapy

from scapy.all import *

def sniff_packets(interface="eth0", packet_count=10):
    packets = sniff(iface=interface, count=packet_count)
    return packets

def analyze_packets(packets):
    summary = []
    for pkt in packets:
        if IP in pkt:
            src = pkt[IP].src
            dst = pkt[IP].dst
            proto = pkt[IP].proto
            summary.append((src, dst, proto))
    return summary

packets = sniff_packets(packet_count=20)
results = analyze_packets(packets)

for r in results:
    print(r)
```
