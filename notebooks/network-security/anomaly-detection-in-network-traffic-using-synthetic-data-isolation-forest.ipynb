# Anomaly Detection in Network Traffic using Synthetic Data + Isolation Forest

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
normal_traffic = np.random.normal(loc=50, scale=10, size=(1000, 5))
anomalies = np.random.normal(loc=100, scale=5, size=(50, 5))
data = np.vstack((normal_traffic, anomalies))
labels = np.array([0]*1000 + [1]*50)

df = pd.DataFrame(data, columns=['duration', 'src_bytes', 'dst_bytes', 'count', 'srv_count'])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

model = IsolationForest(contamination=0.05, random_state=42)
model.fit(X_scaled)

df['anomaly_score'] = model.decision_function(X_scaled)
df['anomaly'] = model.predict(X_scaled)
df['anomaly'] = df['anomaly'].map({1: 0, -1: 1})

print("Detected anomalies:", df['anomaly'].sum())

pca = PCA(n_components=2)
reduced = pca.fit_transform(X_scaled)
df['pca1'] = reduced[:,0]
df['pca2'] = reduced[:,1]

plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='pca1', y='pca2', hue='anomaly', palette=['blue','red'])
plt.title("Anomaly Detection in Synthetic Network Traffic")
plt.show()
```
