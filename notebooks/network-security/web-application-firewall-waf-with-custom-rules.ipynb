# Web Application Firewall (WAF) with Custom Rules

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


## Project overview

**Objective:** Demonstrate defensive HTTP request inspection with understandable, testable custom rules.

**Workflow:** Create synthetic requests, inspect request attributes, apply allow and deny rules, and record the decision with an explanation.

**Expected output:** Request decisions, matched-rule details, and examples of tuning tradeoffs.

**Skills demonstrated:** Application security, HTTP, detection logic, rule tuning, secure architecture

### Limitations

A notebook rule engine is not a production WAF. Real deployments need canonicalization, managed rules, performance testing, observability, and continuous tuning.


```python
from flask import Flask, request, jsonify
import re

app = Flask(__name__)

# Sample malicious patterns
attack_patterns = [
    r"(\b(?:select|insert|drop|union|--)\b)",  # SQL Injection
    r"<script.*?>.*?</script>",  # XSS
    r"(eval\()|(base64_decode\()",  # Remote Code Execution
]

def is_malicious(data):
    for pattern in attack_patterns:
        if re.search(pattern, data, re.IGNORECASE):
            return True
    return False

@app.route('/request', methods=['POST'])
def handle_request():
    user_input = request.form['data']
    if is_malicious(user_input):
        return jsonify({"error": "Malicious input detected!"}), 400
    return jsonify({"message": "Request processed successfully!"})

if __name__ == '__main__':
    app.run(debug=True)
```
