# DNS Spoof Detector

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import socket, pandas as pd
from IPython.display import display

domains = [
    "google.com", "facebook.com", "twitter.com",
    "github.com", "wikipedia.org", "bankofamerica.com"
]

known_good = {
    "google.com": "142.250", "facebook.com": "157.240",
    "twitter.com": "104.244", "github.com": "140.82",
    "wikipedia.org": "208.80", "bankofamerica.com": "171.161"
}

results = []

for domain in domains:
    try:
        ip = socket.gethostbyname(domain)
        match = any(ip.startswith(good_prefix) for good_prefix in [known_good[domain]])
        results.append((domain, ip, "OK" if match else "Suspicious"))
    except:
        results.append((domain, "N/A", "Error"))

df = pd.DataFrame(results, columns=["Domain", "Resolved IP", "Status"])
display(df)
```
