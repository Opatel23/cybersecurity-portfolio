# Port Scanner

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import socket
from datetime import datetime

# Function to scan a range of ports on a given host
def port_scanner(target, start_port=1, end_port=1024):
    print(f"\n🔎 Scanning host: {target}")
    print(f"📅 Start time: {datetime.now()}")
    print(f"⚙️ Ports: {start_port} to {end_port}\n")

    open_ports = []

    for port in range(start_port, end_port + 1):
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(0.5)  # Set timeout to half a second

        result = s.connect_ex((target, port))
        if result == 0:
            print(f"✅ Port {port} is OPEN")
            open_ports.append(port)
        s.close()

    print(f"\n✅ Scan complete! {len(open_ports)} open ports found.")
    return open_ports


target_host = "scanme.nmap.org"  # You can replace this with any IP/domain
open_ports = port_scanner(target_host, 20, 100)
```
