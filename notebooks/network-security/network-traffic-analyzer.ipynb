```python
!pip install scapy

from scapy.all import *
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

def capture_packets(interface="eth0", count=100):
    packets = sniff(iface=interface, count=count)
    return packets

def extract_features(packets):
    features = []
    for pkt in packets:
        if IP in pkt:
            proto = pkt[IP].proto
            src = pkt[IP].src
            dst = pkt[IP].dst
            length = len(pkt)
            features.append([src, dst, proto, length])
    return features

packets = capture_packets(count=200)

data = extract_features(packets)

df = pd.DataFrame(data, columns=["Source", "Destination", "Protocol", "Length"])

proto_count = Counter(df["Protocol"])

plt.bar(proto_count.keys(), proto_count.values())
plt.xlabel("Protocol")
plt.ylabel("Count")
plt.title("Protocol Distribution")
plt.show()

df["Length"].hist(bins=20)
plt.xlabel("Packet Length")
plt.ylabel("Frequency")
plt.title("Packet Size Distribution")
plt.show()
```
