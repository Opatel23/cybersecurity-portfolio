# Detect Suspicious Network Connections (Log Analyzer)

```python
import pandas as pd

data = {
    'timestamp': ['2025-06-03 12:00:00', '2025-06-03 12:01:00', '2025-06-03 12:02:00'],
    'source_ip': ['192.168.1.5', '10.0.0.2', '192.168.1.5'],
    'destination_ip': ['8.8.8.8', '192.168.1.1', '8.8.4.4'],
    'port': [80, 22, 4444]
}

df = pd.DataFrame(data)

suspicious_ports = [4444, 1337, 31337, 6667]
df['suspicious'] = df['port'].isin(suspicious_ports)
print(df[df['suspicious']])
```
