# DNS Fingerprinting & Host Profiling

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
class DNSSnapshot:
    def __init__(self, raw):
        self.raw = raw
        self.data = {'A': set(), 'AAAA': set(), 'MX': set(), 'NS': set(), 'TXT': set(), 'SOA': set()}

    def parse(self):
        for line in self.raw:
            if " : " in line:
                t, v = line.split(" : ", 1)
                if t in self.data:
                    self.data[t].add(v)

    def diff(self, other):
        diff_report = {}
        for key in self.data:
            added = other.data[key] - self.data[key]
            removed = self.data[key] - other.data[key]
            if added or removed:
                diff_report[key] = {"added": list(added), "removed": list(removed)}
        return diff_report

snapshot1 = [
    "A : 96.7.128.198",
    "A : 23.192.228.80",
    "A : 96.7.128.175",
    "A : 23.192.228.84",
    "A : 23.215.0.138",
    "A : 23.215.0.136",
    "AAAA : 2600:1408:ec00:36::1736:7f24",
    "AAAA : 2600:1406:3a00:21::173e:2e66",
    "AAAA : 2600:1408:ec00:36::1736:7f31",
    "AAAA : 2600:1406:bc00:53::b81e:94c8",
    "AAAA : 2600:1406:3a00:21::173e:2e65",
    "AAAA : 2600:1406:bc00:53::b81e:94ce",
    "MX : 0 .",
    "NS : a.iana-servers.net.",
    "NS : b.iana-servers.net.",
    "TXT : v=spf1 -all",
    "TXT : _k2n1y4vw3qtb4skdx9e7dxt97qrmmq9",
    "SOA : ns.icann.org. noc.dns.icann.org. 2025011635 7200 3600 1209600 3600"
]

snapshot2 = [
    "A : 96.7.128.198",
    "A : 23.192.228.84",
    "A : 23.192.228.80",
    "A : 23.215.0.136",
    "A : 23.215.0.138",
    "A : 96.7.128.175",
    "AAAA : 2600:1408:ec00:36::1736:7f31",
    "AAAA : 2600:1406:3a00:21::173e:2e66",
    "AAAA : 2600:1408:ec00:36::1736:7f24",
    "AAAA : 2600:1406:bc00:53::b81e:94ce",
    "AAAA : 2600:1406:3a00:21::173e:2e65",
    "AAAA : 2600:1406:bc00:53::b81e:94c8",
    "MX : 0 .",
    "NS : a.iana-servers.net.",
    "NS : b.iana-servers.net.",
    "TXT : v=spf1 -all",
    "TXT : _k2n1y4vw3qtb4skdx9e7dxt97qrmmq9",
    "SOA : ns.icann.org. noc.dns.icann.org. 2025011635 7200 3600 1209600 3600"
]

snap1 = DNSSnapshot(snapshot1)
snap2 = DNSSnapshot(snapshot2)
snap1.parse()
snap2.parse()
result = snap1.diff(snap2)
print(result)
```
