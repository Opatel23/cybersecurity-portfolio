# Suspicious File Scanner (Offline, No API)

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import pandas as pd
import re

files = [
    {"name": "invoice.pdf", "content": "Invoice details attached"},
    {"name": "setup.exe", "content": "Run this to install the game"},
    {"name": "document.scr", "content": "See attached picture"},
    {"name": "report.docx", "content": "Q3 performance report"},
    {"name": "paycheck.vbs", "content": "Double-click to view salary"},
    {"name": "photo.jpg", "content": "Beach trip 2023"},
    {"name": "update.bat", "content": "System update — click to run"},
    {"name": "README.txt", "content": "Instructions for use"},
]

dangerous_ext = [".exe", ".scr", ".bat", ".vbs", ".js", ".cmd", ".ps1"]
suspicious_terms = ["run", "click", "double-click", "install", "password", "update"]

def check_file(file):
    name = file["name"].lower()
    content = file["content"].lower()
    ext_flag = any(name.endswith(ext) for ext in dangerous_ext)
    content_flag = any(term in content for term in suspicious_terms)
    score = ext_flag + content_flag
    return "Malicious" if score == 2 else "Suspicious" if score == 1 else "Clean"

df = pd.DataFrame(files)
df["Classification"] = df.apply(check_file, axis=1)
df
```
