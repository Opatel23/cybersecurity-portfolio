# Malicious URL Detection (Heuristic-Based, No ML or API)

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import pandas as pd
import re

urls = [
    "http://192.168.0.1/badlogin",
    "https://paypal.com-login.secure-update.co",
    "http://example.com",
    "https://secure-google-account.verification-login.net",
    "https://github.com",
    "http://free-gift-card.win",
    "http://login-facebook.com-security-check.ru"
]

def is_suspicious(url):
    heuristics = [
        lambda u: bool(re.search(r"\d{1,3}(?:\.\d{1,3}){3}", u)),  # IP address in URL
        lambda u: "-" in re.sub(r"https?://", "", u).split("/")[0],  # hyphen in domain
        lambda u: any(word in u.lower() for word in ["login", "secure", "update", "verify", "account", "win"]),
        lambda u: not u.startswith("https"),  # uses HTTP
        lambda u: len(u) > 75  # very long URL
    ]
    score = sum([check(url) for check in heuristics])
    return "Suspicious" if score >= 2 else "Legit"

df = pd.DataFrame(urls, columns=["URL"])
df["Classification"] = df["URL"].apply(is_suspicious)
df
```
