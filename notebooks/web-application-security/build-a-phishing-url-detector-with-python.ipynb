# Build a Phishing URL Detector with Python

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import re, requests

def phishing_score(url):
    score = 0
    keywords = ['login','secure','update','verify','account','paypal','bank']
    tlds = ['.zip','.tk','.ml']
    lookalikes = {'g00gle.com':'google.com','faceb00k.com':'facebook.com'}
    if any(k in url.lower() for k in keywords):
        score += 30
    if any(url.lower().endswith(tld) for tld in tlds):
        score += 20
    for fake, real in lookalikes.items():
        if fake in url.lower():
            score += 40
    try:
        domain = re.findall(r'https?://([^/]+)', url)[0]
        if re.match(r'^[0-9.]+$', domain):
            score += 10
    except:
        score += 10
    return min(score,100)

urls = [
    "http://secure-login-paypal.com",
    "https://g00gle.com",
    "http://192.168.0.1/verify",
    "https://example.com",
    "http://faceb00k.com/login"
]

for u in urls:
    print(u, "-> Risk Score:", phishing_score(u))
```
