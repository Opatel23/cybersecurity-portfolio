# Intermediate Keylogger

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
from datetime import datetime
from IPython.display import display, clear_output
import ipywidgets as widgets
from google.colab import files

log_file = "mock_keylog.txt"
log_data = []

def log_keystrokes(change):
    new_value = change['new']
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    log_entry = f"{timestamp} - {new_value}"
    log_data.append(log_entry)
    with open(log_file, "w") as f:
        for line in log_data:
            f.write(line + "\n")

def show_log(b):
    with open(log_file) as f:
        print(f.read())

def download_log(b):
    files.download(log_file)

input_box = widgets.Text(description='Type:', placeholder='Start typing...')
input_box.observe(log_keystrokes, names='value')

show_button = widgets.Button(description=' Show Logs')
show_button.on_click(show_log)

download_button = widgets.Button(description='⬇ Download Logs')
download_button.on_click(download_log)

clear_output()
display(widgets.HTML("<h3> Simulated Keylogger (Safe in Colab)</h3>"))
display(input_box, show_button, download_button)
print(f"Logging to '{log_file}' – every input change is timestamped.")
```
