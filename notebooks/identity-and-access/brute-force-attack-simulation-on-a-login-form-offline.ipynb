# Brute Force Attack Simulation on a Login Form (Offline)

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import time
import threading
import random

# A basic password list (more complex in real-world applications)
password_list = [
    "123456", "password", "letmein", "qwerty", "password123",
    "admin", "welcome", "guest", "letmein123", "123qwe",
    "iloveyou", "1234abcd", "welcome123", "supersecret",
    "12345", "trustno1", "1234", "abc123", "letmein", "monkey",
    "1q2w3e4r", "qwertyuiop"
]

# Simulate the login process for a user
def attempt_login(user, password, attempt_number):
    print(f"Trying password: {password}")
    time.sleep(random.uniform(0.5, 1.5))  # Simulate a real delay

    if password == "password123":  # Simulate the correct password
        print(f"[+] Password found for '{user}': {password}")
        return True
    else:
        print(f"[-] Incorrect password.")
        return False

# Multi-threading function to simulate multiple login attempts
def multi_threaded_bruteforce(user, password_list):
    failed_attempts = 0
    max_attempts = 3
    lockout_time = 5  # Lockout time in seconds

    for attempt_number, password in enumerate(password_list, start=1):
        if failed_attempts >= max_attempts:
            print(f"[!] Account locked. Too many failed attempts. Please wait for {lockout_time} seconds.")
            time.sleep(lockout_time)
            failed_attempts = 0  # Reset the failed attempts after lockout

        success = attempt_login(user, password, attempt_number)
        if success:
            print(f"[+] Attack succeeded in {attempt_number} attempts.")
            break
        else:
            failed_attempts += 1

# Main function to start the brute force attack
def start_attack():
    user = "admin"
    print(f"Starting brute force attack for user: {user}")

    # Using threading to speed up the process
    threads = []
    for i in range(3):  # Simulate 3 different threads to attempt login at the same time
        thread = threading.Thread(target=multi_threaded_bruteforce, args=(user, password_list))
        threads.append(thread)
        thread.start()

    # Wait for all threads to complete
    for thread in threads:
        thread.join()

if __name__ == "__main__":
    start_attack()
```
