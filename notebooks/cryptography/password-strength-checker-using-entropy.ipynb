# Password Strength Checker using Entropy

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import math
import re

def password_entropy(password):
    charset = 0
    if re.search(r'[a-z]', password): charset += 26
    if re.search(r'[A-Z]', password): charset += 26
    if re.search(r'[0-9]', password): charset += 10
    if re.search(r'[^a-zA-Z0-9]', password): charset += 32
    entropy = len(password) * math.log2(charset) if charset else 0
    return round(entropy, 2)

passwords = ["password", "P@ssw0rd", "123456", "H&7d!Ls9f2@#", "letmein"]
for pwd in passwords:
    print(f"Password: {pwd.ljust(15)} | Entropy: {password_entropy(pwd)} bits")


 ##Entropy (bits)	Strength Level	Crackability
##< 28	Very Weak	Instantly cracked (dictionary/brute force)
28-35	## Weak	A few seconds to minutes
####36–59	Moderate	Hours to days
##60–80	Strong	Years (offline attacks)
##> 80	Very Strong	Decades+ or practically uncrackable
##
```
