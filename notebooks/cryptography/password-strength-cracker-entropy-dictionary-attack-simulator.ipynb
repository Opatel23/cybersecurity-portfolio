# Password Strength Cracker (Entropy + Dictionary Attack Simulator)

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import math
from ipywidgets import Textarea, Button, Output, VBox
from IPython.display import display

sample_wordlist = {"password", "123456", "qwerty", "letmein", "admin", "iloveyou", "welcome", "monkey"}

def shannon_entropy(pwd):
    freq = {}
    for c in pwd:
        freq[c] = freq.get(c, 0) + 1
    probs = [f / len(pwd) for f in freq.values()]
    return -sum(p * math.log2(p) for p in probs)

def dictionary_crack(pwd):
    return pwd.lower() in sample_wordlist

def classify(ent, cracked):
    if cracked:
        return "Very Weak (In Dictionary)"
    elif ent < 2:
        return "Weak"
    elif ent < 3:
        return "Moderate"
    elif ent < 4:
        return "Strong"
    else:
        return "Very Strong"

def analyze_passwords(pwds):
    results = []
    for pwd in pwds:
        ent = shannon_entropy(pwd)
        cracked = dictionary_crack(pwd)
        results.append((pwd, round(ent, 2), cracked, classify(ent, cracked)))
    return results

input_box = Textarea(value='password\nTr0ub4dor&3\ncorrecthorsebatterystaple\n123456\nG00dLuck!',
                     description='Passwords:', layout={'width': '100%', 'height': '150px'})

analyze_btn = Button(description='Analyze', button_style='info')
out = Output()

def on_click(b):
    out.clear_output()
    pwds = [line.strip() for line in input_box.value.splitlines() if line.strip()]
    results = analyze_passwords(pwds)
    with out:
        print(f"{'Password':<30} {'Entropy':<10} {'InDict':<10} {'Strength'}")
        print('-'*60)
        for r in results:
            print(f"{r[0]:<30} {r[1]:<10} {str(r[2]):<10} {r[3]}")

analyze_btn.on_click(on_click)
display(VBox([input_box, analyze_btn, out]))
```
