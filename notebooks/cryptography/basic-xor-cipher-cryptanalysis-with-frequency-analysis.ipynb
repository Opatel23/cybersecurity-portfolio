# Basic XOR Cipher Cryptanalysis with Frequency Analysis

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import string
from collections import Counter
import numpy as np

def xor_decrypt(data, key):
    return bytes([b ^ key for b in data])

def score_text(text):
    freqs = {
        'a': .08167, 'b': .01492, 'c': .02782, 'd': .04253, 'e': .12702, 'f': .02228,
        'g': .02015, 'h': .06094, 'i': .06966, 'j': .00153, 'k': .00772, 'l': .04025,
        'm': .02406, 'n': .06749, 'o': .07507, 'p': .01929, 'q': .00095, 'r': .05987,
        's': .06327, 't': .09056, 'u': .02758, 'v': .00978, 'w': .02360, 'x': .00150,
        'y': .01974, 'z': .00074, ' ': .13000
    }
    text = text.lower()
    counts = Counter(text)
    return sum([counts.get(c, 0) * freqs.get(c, 0) for c in freqs])

cipher_hex = "1b37373331363f78151b7f2b783431333d78397828372d36"
cipher_bytes = bytes.fromhex(cipher_hex)

scores = [(key, xor_decrypt(cipher_bytes, key)) for key in range(256)]
decoded = max(scores, key=lambda x: score_text(x[1].decode(errors='ignore')))
print("Key:", decoded[0])
print("Message:", decoded[1].decode())
```
