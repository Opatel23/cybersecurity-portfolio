# Simple Password Strength Classifier Using Regex & Basic Features

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import re
import pandas as pd
from sklearn.model_selection import train_test_split

data = {
    "password": [
        "123456", "password", "P@ssw0rd!", "letmein", "abc123", "CorrectHorseBatteryStaple",
        "admin", "Qwerty!234", "trustno1", "StrongPass#2023"
    ],
    "label": [0, 0, 1, 0, 0, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

def extract_features(pw):
    return {
        "length": len(pw),
        "has_upper": int(bool(re.search(r"[A-Z]", pw))),
        "has_lower": int(bool(re.search(r"[a-z]", pw))),
        "has_digit": int(bool(re.search(r"\d", pw))),
        "has_special": int(bool(re.search(r"[^A-Za-z0-9]", pw)))
    }

features = df["password"].apply(extract_features).apply(pd.Series)
df = pd.concat([df, features], axis=1)

def classifier(row):
    score = 0
    score += row["length"] >= 8
    score += row["has_upper"]
    score += row["has_lower"]
    score += row["has_digit"]
    score += row["has_special"]
    return int(score >= 4)

df["predicted"] = df.apply(classifier, axis=1)
print(df[["password", "label", "predicted"]])
print((df["label"] == df["predicted"]).mean())
```
