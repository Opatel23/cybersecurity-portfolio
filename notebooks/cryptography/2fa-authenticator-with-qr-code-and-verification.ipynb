# 2FA Authenticator with QR Code and Verification

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
!pip install pyotp qrcode[pil]

import pyotp
import qrcode
from IPython.display import display
from PIL import Image
import io

secret = pyotp.random_base32()
print("Secret key:", secret)

totp = pyotp.TOTP(secret)

uri = totp.provisioning_uri(name="user@example.com", issuer_name="SecureApp")
qr = qrcode.make(uri)

print("Scan this QR code with your authenticator app:")
display(qr)

def verify_code():
    while True:
        code = input("Enter the 6-digit code from your app: ").strip()
        if totp.verify(code):
            print(" Code is valid!")
            break
        else:
            print(" Invalid code, try again.")

verify_code()
```
