# Simple XOR File Encryptor/Decryptor

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
key = b"mysecretkey"

def xor_data(data, key):
    return bytes([b ^ key[i % len(key)] for i, b in enumerate(data)])

with open("/content/input.txt", "wb") as f:
    f.write(b"This is a secret message.")

with open("/content/input.txt", "rb") as f:
    original = f.read()

encrypted = xor_data(original, key)

with open("/content/encrypted.bin", "wb") as f:
    f.write(encrypted)

with open("/content/encrypted.bin", "rb") as f:
    encrypted_read = f.read()

decrypted = xor_data(encrypted_read, key)

with open("/content/decrypted.txt", "wb") as f:
    f.write(decrypted)

print("Done")
```
