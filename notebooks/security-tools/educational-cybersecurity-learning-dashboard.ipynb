```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import hashlib
import base64
import re
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

class CyberSecLearningDashboard:
    def __init__(self):
        self.name = "Educational Cybersecurity Analytics Dashboard"
        self.setup_sample_data()

    def setup_sample_data(self):
        np.random.seed(42)
        dates = [datetime.now() - timedelta(days=x) for x in range(30)]

        self.security_logs = pd.DataFrame({
            'timestamp': dates,
            'event_type': np.random.choice(['login_attempt', 'file_access', 'network_scan', 'admin_action'], 30),
            'source_ip': [f"192.168.1.{np.random.randint(1,255)}" for _ in range(30)],
            'success': np.random.choice([True, False], 30, p=[0.7, 0.3]),
            'severity': np.random.choice(['low', 'medium', 'high'], 30, p=[0.5, 0.3, 0.2])
        })

        self.network_traffic = pd.DataFrame({
            'protocol': np.random.choice(['HTTP', 'HTTPS', 'SSH', 'FTP'], 100),
            'bytes_sent': np.random.exponential(1000, 100),
            'bytes_received': np.random.exponential(2000, 100),
            'suspicious': np.random.choice([True, False], 100, p=[0.15, 0.85])
        })

    def password_strength_analyzer(self, password):
        score = 0
        feedback = []

        if len(password) >= 8:
            score += 1
        else:
            feedback.append("Password should be at least 8 characters long")

        if re.search(r'[A-Z]', password):
            score += 1
        else:
            feedback.append("Add uppercase letters")

        if re.search(r'[a-z]', password):
            score += 1
        else:
            feedback.append("Add lowercase letters")

        if re.search(r'\d', password):
            score += 1
        else:
            feedback.append("Add numbers")

        if re.search(r'[!@#$%^&*(),.?":{}|<>]', password):
            score += 1
        else:
            feedback.append("Add special characters")

        strength_levels = ['Very Weak', 'Weak', 'Fair', 'Good', 'Strong']
        strength = strength_levels[min(score, 4)]

        return {
            'strength': strength,
            'score': f"{score}/5",
            'feedback': feedback
        }

    def hash_demonstration(self, text):
        algorithms = {
            'MD5': hashlib.md5(text.encode()).hexdigest(),
            'SHA-1': hashlib.sha1(text.encode()).hexdigest(),
            'SHA-256': hashlib.sha256(text.encode()).hexdigest(),
            'SHA-512': hashlib.sha512(text.encode()).hexdigest()
        }
        return algorithms

    def base64_encode_decode(self, text, operation='encode'):
        if operation == 'encode':
            encoded = base64.b64encode(text.encode()).decode()
            return f"Encoded: {encoded}"
        else:
            try:
                decoded = base64.b64decode(text).decode()
                return f"Decoded: {decoded}"
            except:
                return "Invalid base64 string"

    def analyze_security_logs(self):
        fig, axes = plt.subplots(2, 2, figsize=(15, 10))

        event_counts = self.security_logs['event_type'].value_counts()
        axes[0,0].pie(event_counts.values, labels=event_counts.index, autopct='%1.1f%%')
        axes[0,0].set_title('Security Event Distribution')

        severity_counts = self.security_logs['severity'].value_counts()
        axes[0,1].bar(severity_counts.index, severity_counts.values,
                     color=['green', 'orange', 'red'])
        axes[0,1].set_title('Event Severity Levels')
        axes[0,1].set_ylabel('Count')

        daily_events = self.security_logs.groupby(self.security_logs['timestamp'].dt.date).size()
        axes[1,0].plot(daily_events.index, daily_events.values, marker='o')
        axes[1,0].set_title('Daily Security Events')
        axes[1,0].set_ylabel('Event Count')
        axes[1,0].tick_params(axis='x', rotation=45)

        success_rate = self.security_logs.groupby('event_type')['success'].mean()
        axes[1,1].bar(success_rate.index, success_rate.values)
        axes[1,1].set_title('Success Rate by Event Type')
        axes[1,1].set_ylabel('Success Rate')
        axes[1,1].tick_params(axis='x', rotation=45)

        plt.tight_layout()
        plt.show()

    def network_traffic_analysis(self):
        fig, axes = plt.subplots(1, 3, figsize=(18, 5))

        protocol_counts = self.network_traffic['protocol'].value_counts()
        axes[0].bar(protocol_counts.index, protocol_counts.values)
        axes[0].set_title('Network Protocol Usage')
        axes[0].set_ylabel('Request Count')

        suspicious_data = self.network_traffic[self.network_traffic['suspicious'] == True]
        axes[1].scatter(self.network_traffic['bytes_sent'], self.network_traffic['bytes_received'],
                       alpha=0.6, label='Normal', color='blue')
        axes[1].scatter(suspicious_data['bytes_sent'], suspicious_data['bytes_received'],
                       alpha=0.8, label='Suspicious', color='red')
        axes[1].set_xlabel('Bytes Sent')
        axes[1].set_ylabel('Bytes Received')
        axes[1].set_title('Network Traffic Pattern Analysis')
        axes[1].legend()

        suspicious_by_protocol = self.network_traffic.groupby('protocol')['suspicious'].mean()
        axes[2].bar(suspicious_by_protocol.index, suspicious_by_protocol.values, color='orange')
        axes[2].set_title('Suspicious Activity Rate by Protocol')
        axes[2].set_ylabel('Suspicious Rate')

        plt.tight_layout()
        plt.show()

    def ip_analysis(self):
        ip_counts = Counter(self.security_logs['source_ip'])
        top_ips = dict(ip_counts.most_common(10))

        plt.figure(figsize=(12, 6))
        plt.bar(range(len(top_ips)), list(top_ips.values()))
        plt.xticks(range(len(top_ips)), list(top_ips.keys()), rotation=45)
        plt.title('Top Source IPs by Activity')
        plt.ylabel('Activity Count')
        plt.xlabel('Source IP')
        plt.tight_layout()
        plt.show()

        return top_ips

    def security_recommendations(self):
        recommendations = [
            "Enable multi-factor authentication (MFA) for all accounts",
            "Implement network segmentation to limit lateral movement",
            "Regular security awareness training for employees",
            "Keep all systems and software updated with latest patches",
            "Use endpoint detection and response (EDR) solutions",
            "Implement zero-trust network architecture",
            "Regular penetration testing and vulnerability assessments",
            "Backup data regularly and test restoration procedures",
            "Monitor and log all network activities",
            "Use strong, unique passwords with a password manager"
        ]

        print("🛡️ CYBERSECURITY BEST PRACTICES:")
        print("=" * 50)
        for i, rec in enumerate(recommendations, 1):
            print(f"{i:2d}. {rec}")

    def run_demo(self):
        print("🔐 CYBERSECURITY LEARNING DASHBOARD")
        print("=" * 50)

        print("\n1. PASSWORD STRENGTH ANALYSIS:")
        test_passwords = ["123456", "Password123", "MyS3cur3P@ssw0rd!"]
        for pwd in test_passwords:
            result = self.password_strength_analyzer(pwd)
            print(f"Password: {pwd[:3]}... | Strength: {result['strength']} ({result['score']})")

        print("\n2. CRYPTOGRAPHIC HASH DEMONSTRATION:")
        sample_text = "SecureData123"
        hashes = self.hash_demonstration(sample_text)
        for algo, hash_val in hashes.items():
            print(f"{algo}: {hash_val[:32]}...")

        print("\n3. BASE64 ENCODING EXAMPLE:")
        original = "sensitive_data"
        encoded_result = self.base64_encode_decode(original, 'encode')
        print(encoded_result)

        print("\n4. SECURITY LOG ANALYSIS:")
        self.analyze_security_logs()

        print("\n5. NETWORK TRAFFIC ANALYSIS:")
        self.network_traffic_analysis()

        print("\n6. IP ADDRESS ACTIVITY ANALYSIS:")
        top_ips = self.ip_analysis()

        print("\n7. SECURITY RECOMMENDATIONS:")
        self.security_recommendations()

dashboard = CyberSecLearningDashboard()
dashboard.run_demo()

print("\n🎯 INTERACTIVE FEATURES:")
print("Try these functions:")
print("• dashboard.password_strength_analyzer('your_password_here')")
print("• dashboard.hash_demonstration('your_text_here')")
print("• dashboard.base64_encode_decode('text_to_encode', 'encode')")
```
