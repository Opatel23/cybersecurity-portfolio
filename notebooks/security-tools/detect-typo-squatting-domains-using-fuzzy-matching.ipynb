# Detect Typo-Squatting Domains Using Fuzzy Matching

```python
!pip install python-Levenshtein

import Levenshtein
import pandas as pd

trusted_domains = [
    'google.com', 'facebook.com', 'youtube.com',
    'amazon.com', 'twitter.com', 'wikipedia.org'
]

suspicious_domains = [
    'g00gle.com', 'faceb00k.com', 'youtubee.com',
    'amaz0n.com', 'twltter.com', 'wikpedia.org',
    'paypal.com', 'goog1e.com', 'wikiopedia.org'
]

def check_similarity(suspicious, trusted_list, threshold=0.8):
    results = []
    for s in suspicious:
        for t in trusted_list:
            ratio = Levenshtein.ratio(s, t)
            if ratio > threshold:
                results.append((s, t, ratio))
    return results

matches = check_similarity(suspicious_domains, trusted_domains, threshold=0.75)

df = pd.DataFrame(matches, columns=['Suspicious Domain', 'Matched Trusted Domain', 'Similarity Score'])
print(df)
```
