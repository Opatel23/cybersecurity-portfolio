# Simple JWT Cracker (Weak Secret Brute-Forcer)

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import jwt
from jwt.exceptions import InvalidSignatureError

class JWTFuzzer:
    def __init__(self, token, wordlist):
        self.token = token
        self.wordlist = wordlist
        self.alg = jwt.get_unverified_header(token)['alg']

    def crack(self):
        for secret in self.wordlist:
            try:
                jwt.decode(self.token, secret, algorithms=[self.alg])
                return secret
            except InvalidSignatureError:
                continue
        return None

payload = {"user": "admin"}
known_secret = "secret"
token = jwt.encode(payload, known_secret, algorithm="HS256")
secrets = ["admin", "secret", "123456", "letmein", "jwtsecret", "password"]

fuzzer = JWTFuzzer(token, secrets)
result = fuzzer.crack()
print(result)
```
