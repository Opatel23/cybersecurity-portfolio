# File Hash Integrity Checker

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
import hashlib
import os

# Create sample files
files = {
    'safe1.txt': 'This is a secure file.',
    'safe2.txt': 'Everything here is okay.',
    'tampered.txt': 'This file was secure but got changed!'
}

for filename, content in files.items():
    with open(filename, 'w') as f:
        f.write(content)

# Generate original hashes
original_hashes = {}
for filename in files:
    with open(filename, 'rb') as f:
        content = f.read()
        original_hashes[filename] = hashlib.sha256(content).hexdigest()

# Simulate tampering
with open('tampered.txt', 'a') as f:
    f.write(' Injected malicious code!')

# Recalculate and compare hashes
for filename in files:
    with open(filename, 'rb') as f:
        new_hash = hashlib.sha256(f.read()).hexdigest()
    status = 'UNTOUCHED' if new_hash == original_hashes[filename] else 'MODIFIED'
    print(f'{filename}: {status}')
```
