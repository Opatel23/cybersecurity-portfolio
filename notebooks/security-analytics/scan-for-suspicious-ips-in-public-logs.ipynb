# Scan for Suspicious IPs in Public Logs

> **Authorized lab use only.** This notebook is an educational demonstration. Run it only with synthetic data or systems you own and are permitted to test.


```python
!pip install requests pandas

import requests
import pandas as pd

log_url = "https://raw.githubusercontent.com/stamparm/ipsum/master/ipsum.txt"
response = requests.get(log_url)
ips = [line.strip() for line in response.text.splitlines() if not line.startswith("#")]
df = pd.DataFrame(ips, columns=["Suspicious_IP"])

print("Top 10 suspicious IPs from public threat feed:")
print(df.head(10))
```
